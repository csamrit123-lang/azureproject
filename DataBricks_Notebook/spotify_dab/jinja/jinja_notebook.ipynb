```python
from jinja2 import Template
```

```python
parameters=[

    {
        "table" : "spotify_cata.silver.factstream",
        "alias" : "factstream",
        "cols" : "factstream.stream_id,factstream.listen_duration"
        
    },
      {
        "table" : "spotify_cata.silver.dimuser",
        "alias" : "dimuser",
        "cols" : "dimuser.user_id,dimuser.user_name",
        "condition" : "dimuser.user_id = factstream.user_id"
        
    },
      {
        "table" : "spotify_cata.silver.dimtrack",
        "alias" : "dimtrack",
        "cols" : "dimtrack.track_id,dimtrack.track_name",
        "condition" : "dimtrack.track_id = factstream.track_id"
        
    }
]
```

```python
query_txt=""" 

    Select
        {% for param in parameters %}
            {{param.cols}}
                {% if not loop.last %}
        , 
                {% endif %}
        {% endfor %}
    from {% for param in parameters %}
            {%if loop.first %}
                {{param['table']}}
            {% endif %}
        {%endfor%}
        {%for param in parameters%}
              {%if not loop.first %}
    left join 
            {{param['table']}} AS {{param['alias']}}
            on
            {{param['condition']}}
            {%endif%}
            {% endfor %}

         


"""
```

```python
jinja_sql_str=Template(query_txt)
query=jinja_sql_str.render(parameters=parameters)
print(query)
```

```python
spark.sql(query).display()  # Check if 'query' is correctly formatted and properly quoted.
```
